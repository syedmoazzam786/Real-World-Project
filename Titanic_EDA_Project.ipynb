
# 🚢 End-to-End Exploratory Data Analysis (EDA) on Titanic Dataset

## 📌 Project Objective
Perform a comprehensive EDA to understand the key factors influencing survival on the Titanic.
This notebook includes:
- Data Loading
- Data Cleaning
- Univariate Analysis
- Bivariate Analysis
- Feature Engineering
- Multivariate Analysis
- Correlation Analysis
- YData Profiling Report


```

# STEP 1: Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ydata_profiling import ProfileReport

sns.set(style='whitegrid')
```

```

# STEP 2: Data Loading

titanic_df = pd.read_csv("Titanic-Dataset.csv")  # Update path if needed

display(titanic_df.head())
display(titanic_df.tail())

print("Shape:", titanic_df.shape)
titanic_df.info()
titanic_df.describe()
```

```

# STEP 3: Data Cleaning

# Missing values
print(titanic_df.isna().sum())

# Age - Median Imputation
median_age = titanic_df['Age'].median()
titanic_df['Age'] = titanic_df['Age'].fillna(median_age)

# Embarked - Mode Imputation
mode_embarked = titanic_df['Embarked'].mode()[0]
titanic_df['Embarked'] = titanic_df['Embarked'].fillna(mode_embarked)

# Cabin - Create Has_Cabin Feature
titanic_df['Has_Cabin'] = titanic_df['Cabin'].notna().astype(int)
titanic_df.drop('Cabin', axis=1, inplace=True)

print("After Cleaning:")
print(titanic_df.isna().sum())
```

```

# STEP 4: Univariate Analysis

fig, axes = plt.subplots(2,3, figsize=(18,12))

sns.countplot(ax=axes[0,0], x='Survived', data=titanic_df)
sns.countplot(ax=axes[0,1], x='Pclass', data=titanic_df)
sns.countplot(ax=axes[0,2], x='Sex', data=titanic_df)
sns.countplot(ax=axes[1,0], x='Embarked', data=titanic_df)
sns.countplot(ax=axes[1,1], x='SibSp', data=titanic_df)
sns.countplot(ax=axes[1,2], x='Parch', data=titanic_df)

plt.tight_layout()
plt.show()

# Numerical
fig, axes = plt.subplots(1,2, figsize=(14,5))
sns.histplot(ax=axes[0], data=titanic_df, x='Age', kde=True)
sns.histplot(ax=axes[1], data=titanic_df, x='Fare', kde=True)
plt.show()
```

```

# STEP 5: Bivariate Analysis

fig, axes = plt.subplots(2,2, figsize=(14,10))

sns.barplot(ax=axes[0,0], x='Pclass', y='Survived', data=titanic_df)
sns.barplot(ax=axes[0,1], x='Sex', y='Survived', data=titanic_df)
sns.barplot(ax=axes[1,0], x='Embarked', y='Survived', data=titanic_df)
sns.barplot(ax=axes[1,1], x='Has_Cabin', y='Survived', data=titanic_df)

plt.tight_layout()
plt.show()

# Age vs Survival
g = sns.FacetGrid(titanic_df, col='Survived', height=5)
g.map(sns.histplot, 'Age', kde=True)
plt.show()

# Fare Outliers
sns.boxplot(y='Fare', data=titanic_df)
plt.show()
```

```

# STEP 6: Feature Engineering

titanic_df['FamilySize'] = titanic_df['SibSp'] + titanic_df['Parch'] + 1
titanic_df['IsAlone'] = 0
titanic_df.loc[titanic_df['FamilySize']==1, 'IsAlone']=1

# Extract Title
titanic_df['Title'] = titanic_df['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)

titanic_df['Title'] = titanic_df['Title'].replace(
['Lady','Countess','Capt','Col','Don','Dr','Major','Rev','Sir','Jonkheer','Dona'],'Rare')

titanic_df['Title'] = titanic_df['Title'].replace('Mlle','Miss')
titanic_df['Title'] = titanic_df['Title'].replace('Ms','Miss')
titanic_df['Title'] = titanic_df['Title'].replace('Mme','Mrs')

sns.barplot(x='FamilySize', y='Survived', data=titanic_df)
plt.show()

sns.barplot(x='IsAlone', y='Survived', data=titanic_df)
plt.show()

sns.barplot(x='Title', y='Survived', data=titanic_df)
plt.show()
```

```

# STEP 7: Multivariate Analysis

sns.catplot(x='Pclass', y='Survived', hue='Sex', data=titanic_df, kind='bar')
plt.show()

sns.violinplot(x='Sex', y='Age', hue='Survived', data=titanic_df, split=True)
plt.show()
```

```

# STEP 8: Correlation Analysis

numeric_cols = titanic_df.select_dtypes(include=np.number)
corr = numeric_cols.corr()

plt.figure(figsize=(12,8))
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.show()
```

```

# STEP 9: YData Profiling Report

profile = ProfileReport(titanic_df, title="Titanic EDA Profiling Report")

profile.to_notebook_iframe()

# Save HTML Report
profile.to_file("Titanic_EDA_Report.html")
```
